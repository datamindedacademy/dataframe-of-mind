```python
import polars as pl
```

```python
pl.read_csv(
    source="../data/measurements.csv",
    has_header=False,
    separator="|",
    infer_schema=True,
    try_parse_dates=True,
)
```

```python
measurement_schema = {
    "id": pl.String,
    "name": pl.String,
    "timestamp": pl.Datetime,
    "blood_pressure": pl.Int32,
    "heart_rate": pl.Int32,
    "temperature": pl.Int32,
    "blood_glucose": pl.Int32,
}
pl.read_csv(
    source="../data/measurements.csv",
    has_header=False,
    separator="|",
    schema=measurement_schema,
)
```

```python
pl.read_csv(
    source="../data/batch_measurements.csv",
    has_header=False,
    separator="|",
    infer_schema=True,
    try_parse_dates=True,
)
```

```python
batch_schema = {
    "name": pl.String,
    "age": pl.Int32,
    "weight": pl.Float32,
    "daily_steps": pl.Int32,
    "timestamp": pl.String,
    "vet_health_check": pl.String,
    "life_stage": pl.String,
}
pl.read_csv(
    source="../data/batch_measurements.csv",
    has_header=False,
    separator="|",
    schema=batch_schema,
).with_columns(pl.col("timestamp").str.to_datetime("%d%m%YT%H:%M:%S")) # this is extra
```
